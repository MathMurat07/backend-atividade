```python
Passo 1: Implementação Monolítica
```

```python
class OnlineStore:

    # Inicializa listas de usuários, produtos e pedidos
    def __init__(self):
        self.users = []
        self.products = []
        self.orders = []

    # Cadastra um usuário
    def register_user(self, user):
        self.users.append(user)

    # Adiciona um produto ao catálogo
    def add_product(self, product):
        self.products.append(product)

    # Realiza um pedido
    def place_order(self, user, product):

        # Verifica se o produto existe
        if product in self.products:

            # Adiciona o pedido na lista
            self.orders.append((user, product))

            return "Order placed successfully"

        # Retorna erro caso o produto não exista
        return "Product not available"


# Exemplo de uso
store = OnlineStore()

# Cadastra usuário
store.register_user("Alice")

# Adiciona produto
store.add_product("Laptop")

# Realiza pedido
print(store.place_order("Alice", "Laptop"))
```

Passo 2: Refatoração para microsserviços


```python

class UserService:

    # Inicializa lista de usuários
    def __init__(self):
        self.users = []

    # Cadastra usuário
    def register_user(self, user):
        self.users.append(user)


class ProductService:

    # Inicializa lista de produtos
    def __init__(self):
        self.products = []

    # Adiciona produto
    def add_product(self, product):
        self.products.append(product)

    # Verifica se o produto está disponível
    def check_availability(self, product):
        return product in self.products


class OrderService:

    # Inicializa lista de pedidos
    def __init__(self):
        self.orders = []

    # Realiza pedido usando o serviço de produtos
    def place_order(self, user, product, product_service):

        # Consulta disponibilidade do produto
        if product_service.check_availability(product):

            # Registra pedido
            self.orders.append((user, product))

            return "Order placed successfully"

        # Produto não encontrado
        return "Product not available"


# Exemplo de uso com microsserviços

# Cria os serviços
user_service = UserService()
product_service = ProductService()
order_service = OrderService()

# Cadastra usuário
user_service.register_user("Alice")

# Adiciona produto
product_service.add_product("Laptop")

# Realiza pedido
print(order_service.place_order("Alice", "Laptop", product_service))
```
